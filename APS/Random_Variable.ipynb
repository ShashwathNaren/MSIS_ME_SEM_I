# Problem set I : Simulating Customer Clicks Using a Bernoulli Random Variable
 
 **Problem statement**

An organization sends promotional emails. Historical data indicate that a customer click
the link with probability p = 0.30.

Define:
$$
X = \begin{cases}
1, & \text{customer clicks} \\
0, & \text{customer does not click}
\end{cases}
$$

* x = 1, customer clicks
* x = 0, customer does not click

Thus,
                
X ~ Bernoulli(O.30)

**tasks**

**Step 1: Identify the random variable**
    
Answer the following:

1. What does $X = 1$ represent?
    * Ans : customer clicks

2. What does $X = 0$ represent?
    * Ans : customer does not click

3. What are the possible values of $X$?
    * Ans: $0$ and $1$

4. Is $$X$$ a discrete or continuous random variable?
    * Ans: discrete random variable

5. Which probability distribution is suitable for $X$?
    * Ans: Bernoulli distribution

**Step 2: Write the theoretical PMF** 

The theoratical PMF for this problem is :

Given :

p = 0.3
1- p = 0.7

Bernoulli distribution formula is as fololows:
        $$ P(X = x) = p^x(1-p)^{1-x} $$

IF $ X = 1 $ then,
$$
\begin{aligned}
P(X = 1) = 0.3^1(0.7)^{1-1}= 0.3 
\end{aligned}
$$

IF $ X = 0 $ then,
$$
\begin{aligned}
P(X = 0) = 0.3^0(0.7)^{1-0}= 0.7 
\end{aligned}
$$
Therefore,

**PMF**
$$
P(X = x) =  \begin{cases}
0.30 & \text{if } X = 1 \\
0.70 & \text{if} X = 0
\end{cases}
$$

**Step 3: Generate outcomes for 20 customers.**
 
Hint:np.random.binomial(n=1,...)  generates Bernoulli observations because only one trial is performed for each customer.

Use code: samples = np.random.binomial(n=1, p=p, size=sample_size)

```python
import random
import numpy as np
p = 0.3
sample_size = 20

samples = np.random.binomial(n=1, p=p, size=sample_size)

print(samples)
```

**Step 7: plot empirical and theoretical PMFs**

```python
import matplotlib.pyplot as plt
import numpy as np

outcomes = [0, 1]
theoretical_pmf = [0.70, 0.30]
empirical_pmf = [0.65, 0.35]

bar_width = 0.25
index = np.arange(len(outcomes))

plt.figure(figsize=(8, 5))

plt.bar(
    index - bar_width / 2,
    theoretical_pmf,
    width=bar_width,
    label="Theoretical PMF",
    color="red",
    edgecolor="black",
)

plt.bar(
    index + bar_width / 2,
    empirical_pmf,
    width=bar_width,
    label="Empirical PMF",
    color="blue",
    edgecolor="black",
)

# Label formatting
plt.xlabel("Outcome (X)", fontsize=12)
plt.ylabel("Probability", fontsize=12)
plt.title("Comparison: Theoretical vs. Empirical Bernoulli PMFs", fontsize=14)
plt.xticks(index, ["0 (Non-click)", "1 (Click)"])
plt.ylim(0, 1.0)
plt.legend()

plt.show()
```

**Step 8: Calculate the theoretical expectation**

Ans:

Theoretical Expectation :
$$
E[X] = \sum \cdot P(X = x)
$$

$$
\begin{aligned}
E[X] &=  (0 \cdot 0.70) + (1 \cdot 0.30) \\
&= 0 + 0.30 \\
&= 0.30
\end{aligned}
$$


**Step 9: Calculate the empirical expectation**

Ans:

Empirical Expectation :

$$
E[X] = \sum \cdot P(X = x)
$$

$$
\begin{aligned}
\bar{x} &= \frac{(1 \cdot 7) + (0 \cdot 13)}{20} \\
&= \frac{7}{20} \\
&= 0.35
\end{aligned}
$$

**Step 10: Calculate the theoretical variance**

Theoretical Variance : 

Ans:

$$
\begin{aligned}
\text{var}(X) &= 0.30 \cdot (1 - 0.30) \\
&= 0.30 \cdot 0.70 \\
&= 0.21
\end{aligned}
$$

**Step 11: Calculate the empirical variance**

Ans:

Empirical Variance :


$$
\begin{aligned}
s^2 &= \frac{\sum (x_i - \bar{x})^2}{n-1} \\
s^2 &= \frac{7 \cdot (1 - 0.35)^2 + 13 \cdot (0 - 0.35)^2}{20 - 1} \\
&= \frac{7 \cdot (0.65)^2 + 13 \cdot (-0.35)^2}{19} \\
&= \frac{7 \cdot (0.4225) + 13 \cdot (0.1225)}{19} \\
&= \frac{2.9575 + 1.5925}{19} \\
&= \frac{4.55}{19} \approx 0.2395
\end{aligned}
$$

**Step 12: Prepare the final result table**

```python
mean_theoretical = 0.30
mean_empirical = 0.35

variance_theoretical = 0.21
variance_empirical = 0.2395

mean_difference = abs(mean_theoretical - mean_empirical)
variance_difference = abs(variance_theoretical - variance_empirical)

print(f"Mean Diffrence : {mean_difference:.2f}")
print(f"Variance Diffrence : {variance_difference:.2f}")
```

```python
import pandas as pd

theoretical_p_no_click = 0.70
theoretical_p_click = 0.30
theoretical_mean = 0.30
theoretical_variance = 0.21

empirical_p_no_click = 0.65
empirical_p_click = 0.35
empirical_mean = 0.35
empirical_variance = 0.2395

abs_diff_p_no_click = abs(theoretical_p_no_click - empirical_p_no_click)
abs_diff_p_click = abs(theoretical_p_click - empirical_p_click)
abs_diff_mean = abs(theoretical_mean - empirical_mean)
abs_diff_variance = abs(theoretical_variance - empirical_variance)

result_table = pd.DataFrame({
    "Measure": ["P(X = 0)", "P(X = 1)", "Expectation", "Variance"],
    "Theoretical value": [theoretical_p_no_click, theoretical_p_click,
                          theoretical_mean, theoretical_variance],
    "Empirical value": [empirical_p_no_click, empirical_p_click,
                        empirical_mean, empirical_variance],
    "Absolute difference": [abs_diff_p_no_click, abs_diff_p_click,
                            abs_diff_mean, abs_diff_variance],
})

print("Final Result Table:\n")
print(result_table.round(4).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

p_click = 0.30
sample_sizes = [10, 100, 1000, 10000]

np.random.seed(42)

rows = []
for sample_size in sample_sizes:
    samples = np.random.binomial(n=1, p=p_click, size=sample_size)
    rows.append({
        "Sample size": sample_size,
        "Empirical P(X = 0)": np.mean(samples == 0),
        "Empirical P(X = 1)": np.mean(samples == 1),
        "Empirical mean": np.mean(samples),
        "Empirical variance": np.var(samples, ddof=1),
    })

sample_size_table = pd.DataFrame(rows)

print("Effect of Sample Size:\n")
print(sample_size_table.round(4).to_string(index=False))
```

# Problem Set 2: ML Use Case: Number of Fraudulent Transactions Flagged

**Problem statement**

A fraud-detection model examines 50 transactions every hour. Assume that each transaction has a 4% probability of being flagged as potentially fraudulent.

Let:

$$X = \text{number of transactions flagged in one hour}$$

Initially assume that individual flagging decisions are independent.

$$X \sim \text{Binomial}(n = 50,\ p = 0.04)$$

The fraud investigation team can examine at most four flagged transactions per hour.

**Find:**

**Task 1:** Simulate the operation of the fraud-detection system for 20,000 independent one-hour periods.

During every simulated hour:

* exactly 50 transactions are examined;
* each transaction has a flagging probability of $0.04$;
* $X$ records the total number of flagged transactions.

*Hint: use the `np.random.binomial` function.*

```python
import numpy as np

n_transactions = 50
p_flag = 0.04
n_hours = 20000

np.random.seed(42)

flagged_per_hour = np.random.binomial(n=n_transactions, p=p_flag, size=n_hours)

print("First 20 hours:", flagged_per_hour[:20])
print("Number of hours simulated:", len(flagged_per_hour))
print("Minimum flagged in an hour:", flagged_per_hour.min())
print("Maximum flagged in an hour:", flagged_per_hour.max())
```
